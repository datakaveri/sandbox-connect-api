# JupyterLite Capabilities Demo

This notebook runs inside JupyterLite, the free browser-only sandbox mode. It uses Pyodide and WebAssembly to run Python in your browser.

Use this notebook to try the main things you can do in JupyterLite: run Python, edit Markdown and code cells, use common Python packages, read local files, create plots, and understand browser-local storage.

JupyterLite does not create Sandbox Connect bookings, Kubernetes notebooks, PVCs, GPU sessions, or managed backend compute.

## 1. Run Python Code

Run the next cell to confirm the browser Python kernel is working.

```python
1 + 1
```

## 2. Use Basic Python and Standard Library Modules

Many standard-library modules work normally in Pyodide.

```python
import math
import json

{
    "sqrt": math.sqrt(16),
    "json_value": json.loads('{"ok": true}')["ok"],
}
```

## 3. Use Pyodide-Supported Python Packages

JupyterLite can use packages included in the active Pyodide distribution, including common data and scientific Python packages.

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "x": [1, 2, 3, 4],
    "y": np.array([2, 4, 6, 8]),
})
df
```

## 4. Create Plots

Browser-compatible visualization packages such as matplotlib can render output inside the notebook.

```python
import matplotlib.pyplot as plt

df.plot(x="x", y="y", marker="o", title="JupyterLite Plot Demo")
plt.show()
```

## 5. Read Files From the JupyterLite Workspace

This demo includes `sample-data.csv`. JupyterLite files can be read from Python after they are present in the browser workspace.

```python
import pandas as pd

sample = pd.read_csv("sample-data.csv")
sample
```

## 6. Upload and Download Files

Upload files from the JupyterLite file browser when you want to work with local `.ipynb`, `.csv`, `.txt`, or other browser-supported files.

Download notebooks and files from the JupyterLite file browser when you need a copy outside this browser profile. This is important because JupyterLite storage is local to the browser.

## 7. Install Pure-Python Packages at Runtime

The next cell is optional and network-dependent. It uses `%pip install`, which JupyterLite routes through Pyodide package tooling. It works best for pure-Python packages or packages with Pyodide-compatible wheels.

If your browser or network blocks package downloads, this cell can fail even though the rest of JupyterLite is working.

```python
%pip install snowballstemmer

import snowballstemmer

stemmer = snowballstemmer.stemmer("english")
stemmer.stemWords("running runs runner".split())
```

## 8. Browser-Local Persistence

Files and settings are stored in browser-local storage. They usually remain available when you return in the same browser profile on the same device.

They are not synced to Sandbox Connect storage, Kubernetes PVCs, or user profiles. Clearing browser site data, using a different browser, or using a private/incognito profile can make files disappear.

## 9. Browser Sandbox Limitations

JupyterLite is not a full server-backed notebook. It does not provide terminal access, GPU access, system package installation, native binaries, arbitrary OS access, managed PVC storage, or long-running backend processes.

The next cell intentionally tries an OS-style operation and catches the expected browser limitation.

```python
import subprocess

try:
    subprocess.run(["echo", "hello"], capture_output=True, text=True)
except Exception as exc:
    print("Expected browser limitation:", type(exc).__name__, str(exc))
```
