## 5. Process Large Databanks with the File Iterator

The [DatabankFileIterator](cci:2://file:///d:/IUDX/tgdex/databank-download-sdk/databank_download_widget/manager.py:61:0-117:70) is a context manager that lets you process files one by one, which is ideal for very large databanks or when you have limited storage.

```python
databank_id = "your_databank_id"  # Replace with a real databank ID

try:
    with DatabankFileIterator(databank_id) as databank:
        for file_handler in databank:
            print(f"Processing file: {file_handler.key}")
            
            # Download the file
            local_file_path = file_handler.download()
            
            if local_file_path:
                print(f"  Downloaded to: {local_file_path}")
                
                # --- Add your file processing logic here ---
                
                # Clean up the downloaded file to save space
                print(f"  Cleaning up {file_handler.key}...")
                file_handler.cleanup()
                print("  Done.")
                
except ConnectionError as e:
    print(f"An error occurred: {e}")
```
