# 📦 Package Installation Guidelines


## Installing packages using pip - For instance: PyTorch

````python
pip install torch torchvision torchaudio
````

For more information on how to install python packages, refer to this link: https://packaging.python.org/en/latest/tutorials/installing-packages/

## Installing multiple using a requirements.txt and pip

Create a requirements.txt file and save it

You can manage all your dependencies in a requirements.txt file. Here’s an example you might use in an ML project:


```txt
torch
torchvision
torchaudio
lightgbm
xgboost
scikit-learn
pandas
matplotlib
```


### To install all packages listed in requirements.txt, run:

````python
pip install -r requirements.txt
````

# Check Installed Packages

## To list all installed packages:

````python
pip list
````

## Check Details for a Specific Package (e.g., PyTorch)

````python
pip show torch
````
